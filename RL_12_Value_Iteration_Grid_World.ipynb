### **Write Python code for Value iterative algorithm in a Deterministic Grid World**

Define the MDP

```python
# Define states and actions
states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

# Define transitions
transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

# Define parameters
gamma = 0.9
normal_reward = -1
goal_reward = 100

# Initialize state values
V = {'S0': 0, 'S1': 0, 'G': 0}

print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Normal Reward:", normal_reward)
print("Goal Reward:", goal_reward)
print("\nTransition Information:")

for state in ['S0', 'S1']:
    for action in actions:
        next_state = transitions[state][action]
        reward = goal_reward if next_state == 'G' else normal_reward
        print(state, action, "->", next_state, "Reward:", reward)
```

Implement the Value Iteration Algorithm

```python
theta = 0.001
iteration = 0

print("Iteration-wise State Values")
print("--------------------------------")

while True:
    new_V = V.copy()
    delta = 0

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state = transitions[state][action]

            reward = goal_reward if next_state == 'G' else normal_reward

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    for state in ['S0', 'S1']:
        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    iteration += 1

    print(f"Iteration {iteration}: S0 = {V['S0']:.2f}, S1 = {V['S1']:.2f}, Delta = {delta:.4f}")

    if delta < theta:
        break

print("\nValue Iteration Converged!")
```

Find the Optimal Policy

```python
print("Final Action Value Table")
print("-----------------------------")

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state = transitions[state][action]

        reward = goal_reward if next_state == 'G' else normal_reward

        Q[state][action] = reward + gamma * V[next_state]

        print(state, action, ":", round(Q[state][action], 2))
```
