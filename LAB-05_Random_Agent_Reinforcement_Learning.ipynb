Import Required Libraries

```python
import gymnasium as gym
import random
import numpy as np
```

Create the Random Agent Function

```python
def random_agent(env, episodes=10):
    total_rewards = []

    for episode in range(episodes):
        state, info = env.reset()
        total_reward = 0
        done = False

        if episode == 0:
            print("Initial State:", state)
            print("Action Space:", env.action_space)

        while not done:
            action = env.action_space.sample()

            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            state = next_state

            done = terminated or truncated

        total_rewards.append(total_reward)

        print("Episode:", episode + 1, "Total Reward:", total_reward)

    print("\nTotal Reward:", sum(total_rewards))
    print("Average Reward:", np.mean(total_rewards))

    return total_rewards
```

Implement CartPole Environment

```python
env = gym.make("CartPole-v1")

print("CartPole Environment")
print("Observation Space:", env.observation_space)
print("Action Space:", env.action_space)

cartpole_rewards = random_agent(env, episodes=10)

env.close()
```

Implement MountainCar Environment

```python
env = gym.make("MountainCar-v0")

print("MountainCar Environment")
print("Observation Space:", env.observation_space)
print("Action Space:", env.action_space)

mountaincar_rewards = random_agent(env, episodes=10)

env.close()
```

Implement FrozenLake Environment

```python
env = gym.make("FrozenLake-v1")

print("FrozenLake Environment")
print("Observation Space:", env.observation_space)
print("Action Space:", env.action_space)

frozenlake_rewards = random_agent(env, episodes=10)

env.close()
```

Compare All Three Environments

```python
print("FINAL COMPARISON OF RANDOM AGENT")

print("\nCartPole")
print("Total Reward:", sum(cartpole_rewards))
print("Average Reward:", np.mean(cartpole_rewards))

print("\nMountainCar")
print("Total Reward:", sum(mountaincar_rewards))
print("Average Reward:", np.mean(mountaincar_rewards))

print("\nFrozenLake")
print("Total Reward:", sum(frozenlake_rewards))
print("Average Reward:", np.mean(frozenlake_rewards))
```
