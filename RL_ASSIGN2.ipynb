```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays and exploration probability
N_play = 10
epsilon = 0.2

# Part B - Initialize N, R and Q tables
# N = number of times each machine is selected
# R = total reward obtained from each machine
# Q = estimated average reward of each machine

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

print("=== Epsilon-Greedy Multi-Armed Bandit ===")
print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)

# Part C, D - Decision, Reward Collection and Q-value Update
for play in range(1, N_play + 1):

    print("\nPlay", play)

    # Generate random number
    r = random.random()
    print("Random number:", round(r, 4))

    # Epsilon-Greedy decision
    if r < epsilon:
        # Exploration
        selected_machine = random.choice(machines)
        print("Decision: Exploration")
    else:
        # Exploitation
        max_q = max(Q.values())

        # If multiple machines have the same Q-value,
        # randomly select one among them
        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        print("Decision: Exploitation")

    print("Selected machine:", selected_machine)

    # Part C - Reward Collection
    reward = float(input("Enter reward: "))

    # Part D - Update N and R
    N[selected_machine] += 1
    R[selected_machine] += reward

    # Incremental sample-average formula:
    # Q(a) = Q(a) + [Reward - Q(a)] / N(a)

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    print("Updated N:", N)
    print("Updated R:", R)
    print("Updated Q:", Q)


# Part E - Display Final Results
print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)

print("\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Value)")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t\t"
        f"{Q[machine]:.2f}"
    )

print("\nAnalysis:")

for machine in machines:
    print(
        f"Machine {machine}: Played {N[machine]} times, "
        f"total reward = {R[machine]:.2f}, "
        f"estimated Q-value = {Q[machine]:.2f}"
    )

best_machine = max(Q, key=Q.get)

print(
    f"\nBest machine based on the final Q-value: "
    f"Machine {best_machine}"
)
```
