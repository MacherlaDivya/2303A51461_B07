```python

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']
gamma = 0.9
theta = 0.0001
goal_reward = 100

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },
    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', goal_reward)
    }
}

print("STATE, ACTION, REWARD AND TRANSITIONS")
for s in ['S0', 'S1']:
    for a in actions:
        ns, r = transitions[s][a]
        print(s, a, "->", ns, "Reward:", r)

policy = {'S0': 'Up', 'S1': 'Up'}
V = {'S0': 0, 'S1': 0, 'G': 0}

iteration = 0

while True:
    iteration += 1

    # Policy Evaluation
    while True:
        delta = 0
        new_V = V.copy()

        for s in ['S0', 'S1']:
            ns, reward = transitions[s][policy[s]]
            new_V[s] = reward + gamma * V[ns]
            delta = max(delta, abs(new_V[s] - V[s]))

        V = new_V

        if delta < theta:
            break

    print("\nPolicy Iteration:", iteration)
    print("Policy:", policy)
    print("V(S0):", round(V['S0'], 4))
    print("V(S1):", round(V['S1'], 4))

    # Calculate Action Values
    Q = {}

    for s in ['S0', 'S1']:
        Q[s] = {}
        for a in actions:
            ns, reward = transitions[s][a]
            Q[s][a] = reward + gamma * V[ns]

    # Policy Improvement
    new_policy = {}

    for s in ['S0', 'S1']:
        new_policy[s] = max(Q[s], key=Q[s].get)

    print("Improved Policy:", new_policy)

    # Policy Stability
    if new_policy == policy:
        break

    policy = new_policy

# Final Results
print("\nFINAL OPTIMAL STATE VALUES")
for s in states:
    print(s, ":", round(V[s], 4))

print("\nFINAL ACTION VALUE TABLE")
print("State\tUp\tDown\tLeft\tRight")

for s in ['S0', 'S1']:
    print(s, end="\t")
    for a in actions:
        print(round(Q[s][a], 4), end="\t")
    print()

print("\nOPTIMAL POLICY")
for s in ['S0', 'S1']:
    print(s, ":", policy[s])

print("G : Terminal")

# Bellman Optimality Verification
print("\nBELLMAN OPTIMALITY VERIFICATION")

for s in ['S0', 'S1']:
    max_q = max(Q[s].values())
    print(s, "V*(s) =", round(V[s], 4),
          "Max Q*(s,a) =", round(max_q, 4))
    print("Verified:", abs(V[s] - max_q) < theta)

print("G V*(G) = 0")
print("\nOPTIMAL PATH: S0 -> S1 -> G")
```
