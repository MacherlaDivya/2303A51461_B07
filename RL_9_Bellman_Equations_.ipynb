### Calculation of State and Action Values Using Bellman Expectation Equations for stochastic policy and deterministic environment

```python
gamma=0.9
V0=71.819/(1-0.0729)
V1=89.9+0.09*V0

Q0_down=-1+gamma*V1
Q0_right=-1
Q1_right=100
Q1_up=-1+gamma*V0

V0_check=0.9*Q0_down+0.1*Q0_right
V1_check=0.9*Q1_right+0.1*Q1_up

print("State Values:")
print("V(S0) =",round(V0,2))
print("V(S1) =",round(V1,2))

print("\nAction Values:")
print("Q(S0,Down) =",round(Q0_down,2))
print("Q(S0,Right) =",round(Q0_right,2))
print("Q(S1,Right) =",round(Q1_right,2))
print("Q(S1,Up) =",round(Q1_up,2))

print("\nVerification:")
print("V(S0) =",round(V0_check,2))
print("V(S1) =",round(V1_check,2))
print("Verification successful:",abs(V0-V0_check)<1e-9 and abs(V1-V1_check)<1e-9)
```
