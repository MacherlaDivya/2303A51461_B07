### Write Python code for Bellman Optimality and Value Iteration in a Deterministic Grid World

```python

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]
gamma = 0.9
theta = 0.001
rewards = {"normal": -1, "wall": -1, "goal": 10}
terminal = "G"

transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 10)
    }
}

print("1. MDP INFORMATION")
print("States:", states)
print("Actions:", actions)
print("Rewards:", rewards)
print("Discount factor:", gamma)
print("Terminal state:", terminal)
print("\n2. TRANSITION VERIFICATION")

for s in states:
    if s != terminal:
        for a in actions:
            ns, r = transitions[s][a]
            print(f"{s} --{a}--> {ns}, Reward={r}")

V = {s: 0.0 for s in states}
print("\n3. INITIAL STATE VALUES")
print(V)

print("\n4. VALUE ITERATION")
iteration = 0

while True:
    newV = V.copy()
    for s in states:
        if s != terminal:
            newV[s] = max(
                r + gamma * V[ns]
                for ns, r in transitions[s].values()
            )
    delta = max(abs(newV[s] - V[s]) for s in states)
    iteration += 1
    V = newV
    print(f"Iteration {iteration}: V={V}, Delta={delta:.6f}")
    if delta < theta:
        break

print("\n5. FINAL OPTIMAL STATE VALUES")
for s in states:
    print(f"V*({s}) = {V[s]:.6f}")

print("\n6. ACTION VALUES Q*(s,a)")
Q = {}

for s in states:
    if s != terminal:
        Q[s] = {}
        for a in actions:
            ns, r = transitions[s][a]
            Q[s][a] = r + gamma * V[ns]
            print(f"Q*({s},{a}) = {Q[s][a]:.6f}")

print("\n7. OPTIMAL POLICY")
for s in states:
    if s != terminal:
        best = max(Q[s].values())
        policy = [a for a in actions if abs(Q[s][a] - best) < 1e-9]
        print(f"Policy({s}) = {policy}")

print("\n8. BELLMAN OPTIMALITY VERIFICATION")
for s in states:
    if s != terminal:
        maxQ = max(Q[s].values())
        print(f"{s}: V*={V[s]:.6f}, Max Q*={maxQ:.6f}, "
              f"Verified={abs(V[s]-maxQ) < 1e-9}")
```
