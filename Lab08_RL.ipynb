### **Calculate State Value and Action Value for each state**

```python
gamma=0.9
states=["S0","S1","S3","S4"]
positions={"S0":(0,0),"S1":(0,1),"G":(0,2),"X":(1,0),"S3":(1,1),"S4":(1,2)}
actions=["Up","Down","Left","Right"]
policy={"S0":"Right","S1":"Down","S3":"Right","S4":"Up"}
grid=[["S0","S1","G"],["X","S3","S4"]]

def step(state,action):
    r,c=positions[state]
    moves={"Up":(-1,0),"Down":(1,0),"Left":(0,-1),"Right":(0,1)}
    dr,dc=moves[action]
    nr,nc=r+dr,c+dc
    if not (0<=nr<2 and 0<=nc<3):
        return state,-1,False
    next_state=grid[nr][nc]
    if next_state=="X":
        return next_state,-5,True
    if next_state=="G":
        return next_state,10,True
    return next_state,-1,False

print("Environment created successfully")
print("Policy:",policy)
```

```python
def state_value(state):
    total=0
    discount=1
    while state not in ["G","X"]:
        action=policy[state]
        next_state,reward,done=step(state,action)
        total+=discount*reward
        if done:
            break
        discount*=gamma
        state=next_state
    return total

V={}
for s in states:
    V[s]=state_value(s)
    print(f"V({s}) = {V[s]:.2f}")
```

```python
def action_value(state,action):
    next_state,reward,done=step(state,action)
    if done:
        return reward
    return reward+gamma*V[next_state]

for a in actions:
    print(f"Q(S0,{a}) = {action_value('S0',a):.3f}")
```

```python

print("STATE VALUE FUNCTION")
for s in states:
    print(s,":",round(V[s],3))

print("\nACTION VALUE FUNCTION")
for s in states:
    for a in actions:
        print(f"Q({s},{a}) = {action_value(s,a):.3f}")
```
